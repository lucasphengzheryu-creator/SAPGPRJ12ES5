```python
def get_valid_input(prompt_text, options_dict=None, is_api=False):
    while True:
        if is_api:
            try:
                val = int(input(prompt_text))
                if val < 0:
                    print("ERROR: API MUST BE A NON-NEGATIVE INTEGER. PLEASE TRY AGAIN.\n")
                    continue
                return val
            except ValueError:
                print("ERROR: INVALID API VALUE. MUST BE AN INTEGER. PLEASE TRY AGAIN.\n")
        else:
            print(prompt_text)
            choice = input("Enter choice (1-3): ").strip()
            if choice in options_dict:
                return options_dict[choice]
            print("ERROR: INVALID SELECTION. PLEASE TRY AGAIN.\n")

def haze():
    print("Haze Health Risk Assessment")
    print("-----------------------------")

    # 1. Collect and Validate API
    api = get_valid_input("Enter air pollutant index (API): ", is_api=True)
    print("-" * 30)

    # Classify Air Quality early to determine skip logic
    if 0 <= api <= 50:
        aq_category = "Good"
    elif 51 <= api <= 100:
        aq_category = "Moderate"
    elif 101 <= api <= 200:
        aq_category = "Unhealthy"
    elif 201 <= api <= 300:
        aq_category = "Very Unhealthy"
    else:
        aq_category = "Hazardous"

    # Check if API is healthy (<= 100) or critical (> 300)
    skip_profile = (api <= 100) or (api > 300)

    if not skip_profile:
        # Collect and Validate Age Group
        age_options = {"1": "Under 18", "2": "18–59", "3": "60 and above"}
        age_prompt = "\nSelect Age Group:\n1. Under 18\n2. 18-59\n3. 60 and above"
        age_group = get_valid_input(age_prompt, age_options)
        print("-" * 30)

        # Collect and Validate Outdoor Exposure
        exposure_options = {"1": "Low", "2": "Moderate", "3": "High"}
        exposure_prompt = "\nSelect Outdoor Exposure Level:\n1. Low\n2. Moderate\n3. High"
        outdoor_exposure = get_valid_input(exposure_prompt, exposure_options)
        print("-" * 30)

        # Collect and Validate Health Condition
        health_options = {"1": "Yes", "2": "No", "3": "Prefer not to say"}
        health_prompt = "\nExisting respiratory or heart condition?\n1. Yes\n2. No\n3. Prefer not to say"
        health_condition = get_valid_input(health_prompt, health_options)
        print("-" * 30)

        # Collect and Validate Gender
        gender_options = {"1": "Male", "2": "Female", "3": "Prefer not to say"}
        gender_prompt = "\nSelect Gender:\n1. Male\n2. Female\n3. Prefer not to say"
        gender = get_valid_input(gender_prompt, gender_options)
        print("-" * 30)
    else:
        # Set defaults if profile questions are skipped
        age_group = "Not Provided"
        outdoor_exposure = "Not Provided"
        health_condition = "Not Provided"
        gender = "Not Provided"

    # Stage 2: Assess Personal Precautions
    precaution_assessment = "General precautions apply."
    factors = []
    actions = []

    if api <= 50:
        precaution_assessment = "Routine activities."
        actions.extend([
            "Enjoy outdoor activities.",
            "Monitor local air quality updates."
        ])
    elif 51 <= api <= 100:
        precaution_assessment = "General precautions recommended."
        actions.extend([
            "Normal outdoor activities are acceptable for most individuals.",
            "Monitor air quality levels regularly."
        ])
    else:  # API above 100
        precaution_assessment = "Extra caution recommended."
        actions.extend([
            "Reduce prolonged outdoor activities.",
            "Avoid strenuous exercise outdoors.",
            "Reduce exposure to polluted outdoor air.",
            "Follow official air-quality and health advisories.",
            "Seek medical advice if concerning respiratory symptoms develop."
        ])

        if not skip_profile:
            if outdoor_exposure == "High":
                factors.append("High outdoor exposure expected")
            if age_group in ["Under 18", "60 and above"]:
                factors.append(f"Sensitive age group ({age_group})")
            if health_condition == "Yes":
                factors.append("Pre-existing respiratory or heart condition reported")

    if api > 300:
        precaution_assessment = "Extreme caution required."
        actions = [
            "Avoid all outdoor physical activities.",
            "Stay indoors as much as possible.",
            "Keep windows and doors closed.",
            "Use air purifiers if available and wear particulate masks (e.g., N95) if outdoor transit is necessary.",
            "Strictly follow official public health advisories."
        ]

    # Display Output Results
    print("\nHaze Health Risk Assessment")
    print(f"Reported API:             {api}")
    print(f"Air quality:              {aq_category}")
    print("\nUser profile:")
    print(f"  Age group:              {age_group}")
    print(f"  Outdoor exposure:       {outdoor_exposure}")
    print(f"  Existing health condition:{health_condition}")
    print(f"  Gender:                 {gender}")
    print(f"\nPrecaution assessment:    {precaution_assessment}")

    if factors:
        print("\nFactors prompting additional caution:")
        for factor in factors:
            print(f"  - {factor}")

    print("\nRecommended actions:")
    for action in actions:
        print(f"  - {action}")

# Run the function
haze()
```
